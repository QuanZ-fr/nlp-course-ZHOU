## The data

`data/moby.txt` is the full text of *Moby Dick* (Herman Melville, 1851). `moby_tokens` are its NLTK word tokens and `text1` the corresponding `nltk.Text` object.

## Setup

```python
import os, re, urllib.request
import nltk
import pandas as pd
import numpy as np
for resource in ["punkt", "punkt_tab", "stopwords", "wordnet"]:
    nltk.download(resource, quiet=True)
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords

def local_file(name):
    path = os.path.join("data", name)
    if not os.path.exists(path):
        os.makedirs("data", exist_ok=True)
        urllib.request.urlretrieve(f"https://raw.githubusercontent.com/ababacaryoro/nlp-course/main/assignments/1/data/{name}", path)
    return path

with open(local_file("moby.txt"), encoding="utf-8") as f:
    moby_raw = f.read()
moby_tokens = word_tokenize(moby_raw)
text1 = nltk.Text(moby_tokens)
print(f"{len(moby_raw)} characters, {len(moby_tokens)} tokens")
print(moby_raw[:200])
```

## Examples

```python
def example_one():
    """Number of tokens."""
    return len(moby_tokens)

def example_two():
    """Number of distinct tokens."""
    return len(set(moby_tokens))

def example_three():
    """Number of distinct tokens after lemmatizing verbs."""
    lemmatizer = WordNetLemmatizer()
    return len(set(lemmatizer.lemmatize(w, "v") for w in text1))

print(example_one(), example_two(), example_three())
```

## Question 1 (1 point)

Lexical diversity: number of distinct tokens divided by the number of tokens. *Returns a float.*

```python
def question_one():
    """Lexical diversity: distinct tokens divided by all tokens."""
    return len(set(moby_tokens)) / len(moby_tokens)

print(question_one())   # about 0.08
```

## Question 2 (1 point)

Percentage of tokens equal to `whale` or `Whale`. *Returns a float (e.g. 0.5 for 0.5 %).*

```python
def question_two():
    """Percentage of tokens equal to 'whale' or 'Whale'."""
    whale_count = sum(token in ("whale", "Whale") for token in moby_tokens)
    return 100 * whale_count / len(moby_tokens)

print(question_two())
```

## Question 3 (1 point)

The 20 most frequent tokens and their frequency. *Returns a list of 20 `(token, frequency)` tuples sorted by decreasing frequency.*

```python
def question_three():
    """The 20 most frequent tokens and their frequencies."""
    return nltk.FreqDist(moby_tokens).most_common(20)

print(question_three())
```

## Question 4 (1 point)

Tokens longer than 5 characters with a frequency above 150. *Returns an alphabetically sorted list of tokens.*

```python
def question_four():
    """Alphabetically sorted tokens longer than 5 chars with frequency above 150."""
    frequency = nltk.FreqDist(moby_tokens)
    return sorted(token for token, count in frequency.items()
                  if len(token) > 5 and count > 150)

print(question_four())
```

## Question 5 (1 point)

The longest token and its length. *Returns a `(token, length)` tuple.*

```python
def question_five():
    """The longest token and its length."""
    longest = max(moby_tokens, key=len)
    return longest, len(longest)

print(question_five())
```

## Question 6 (1 point)

Alphabetic tokens (`isalpha()`) with a frequency above 2000. *Returns a list of `(frequency, word)` tuples sorted by decreasing frequency.*

```python
def question_six():
    """Alphabetic tokens with frequency above 2000, sorted by frequency."""
    frequency = nltk.FreqDist(token for token in moby_tokens if token.isalpha())
    return [(count, token) for token, count in frequency.most_common()
            if count > 2000]

print(question_six())
```

## Question 7 (1 point)

Average number of tokens per sentence (`sent_tokenize` for the sentences). *Returns a float.*

```python
def question_seven():
    """Average number of word tokens per sentence."""
    sentences = sent_tokenize(moby_raw)
    return len(moby_tokens) / len(sentences)

print(question_seven())
```

## Question 8 (1 point)

The 10 most frequent lowercase alphabetic tokens after removing the English stop words. *Returns a list of 10 `(word, frequency)` tuples sorted by decreasing frequency.*

```python
def question_eight():
    """Ten most frequent lowercase alphabetic tokens after stop-word removal."""
    stop_words = set(stopwords.words("english"))
    tokens = (token.lower() for token in moby_tokens
              if token.isalpha() and token.lower() not in stop_words)
    return nltk.FreqDist(tokens).most_common(10)

print(question_eight())
```

## Question 9 (1 point)

The 10 most frequent Porter stems of the alphabetic tokens. *Returns a list of 10 `(stem, frequency)` tuples sorted by decreasing frequency.*

```python
def question_nine():
    """Ten most frequent Porter stems of alphabetic tokens."""
    stemmer = PorterStemmer()
    stems = (stemmer.stem(token) for token in moby_tokens if token.isalpha())
    return nltk.FreqDist(stems).most_common(10)

print(question_nine())
```

## Question 10 (1 point)

Preprocess the first 1 000 characters of the novel: tokenize, lowercase, keep alphabetic tokens, remove stop words, lemmatize (default noun POS). *Returns the list of tokens.*

```python
def question_ten():
    """Preprocess the first 1000 characters using default noun lemmatization."""
    text = moby_raw[:1000]
    stop_words = set(stopwords.words("english"))
    lemmatizer = WordNetLemmatizer()
    tokens = word_tokenize(text)
    return [lemmatizer.lemmatize(token.lower()) for token in tokens
            if token.isalpha() and token.lower() not in stop_words]

r = question_ten()
print(len(r), r[:20])
```
